# Preprocessing Data

* `OneHotEncoder` (sklearn)
* `get_dummies` (pandas)

```python
# Print disribution of iris dataset
from sklearn.datasets import load_iris
iris = load_iris()
X = iris.data
y = iris.target

import matplotlib.pyplot as plt
plt.xkcd()
plt.hist(y)
plt.show()
```

```python
# Get dummies for target names
import pandas as pd

#convert to dataframe
iris_df = pd.DataFrame(iris.data, columns=iris.feature_names)
print(iris_df.head())

iris_dummies = pd.get_dummies(iris_df, drop_first=True)
print(iris_dummies.dtypes)
```

```python
# Add dummies to iris dataset

iris_df = pd.concat([iris_df, iris_dummies], axis=1)
print(iris_df.head())

print(iris_dummies.columns)
```
