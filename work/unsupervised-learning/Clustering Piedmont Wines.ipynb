# Clustering Piedmont Wine data using KMeans

```python
from sklearn.datasets import load_wine
dataset = load_wine()
```

```python
from sklearn.cluster import KMeans

# Use KMeans to cluster data into 3 clusters
kmeans = KMeans(n_clusters=3, random_state=0, n_init='auto')
labels = kmeans.fit_predict(dataset.data)

# Print distance from each point to its centroid, known as _inertia_
print(kmeans.inertia_)
```

```python
# Create a cross-tabulation table of the predicted labels and target values
import pandas as pd

# map the target class_x to Barbera, Barolo, and Grignolino
dataset.target_names = ['Barbera', 'Barolo', 'Grignolino']

varieties = [dataset.target_names[i] for i in dataset.target]
print(varieties)
```

```python
labels = kmeans.labels_

# check the predicted vs actual target values
df = pd.DataFrame({'labels': labels, 'varieties': varieties})
print(df)
```

```python
ct = pd.crosstab(df['labels'], df['varieties'])

print(ct)
```

## Use Pipeline to scale and cluster the data in one step

```python
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline

# Create a pipeline that scales the data then performs KMeans
pipeline = make_pipeline(StandardScaler(), KMeans(n_clusters=3, random_state=0, n_init='auto'))
pipeline.fit(dataset.data)

labelsScaled = pipeline.predict(dataset.data)
```

```python
# Recall the cross-tabulation table of the predicted labels (without scaling)
print(ct)
```

```python
dfScaled = pd.DataFrame({'labels': labelsScaled, 'varieties': varieties})
ctScaled = pd.crosstab(dfScaled['labels'], dfScaled['varieties'])

# Rhe cross-tabulation tables of the predicted labels (with scaling)
print(ctScaled)
```
