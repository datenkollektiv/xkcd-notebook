# Intrinsic Dimensions

Is the number of dimensions that are required to describe the data (in an approximate way).

The essential number of dimensions to describe the data is the number of degrees of freedom of the data.

```python
from sklearn.datasets import load_iris
iris = load_iris()
```

```python
# Perform PCA on iris data
from sklearn.decomposition import PCA

# Create a PCA model with 2 PCA features
pca = PCA(n_components=2)

pca.fit(iris.data)
transformed = pca.transform(iris.data)
print(transformed.shape)
```

```python
from matplotlib import pyplot as plt
plt.xkcd()

xs = transformed[:,0]
ys = transformed[:,1]

plt.title('PCA on Iris Dataset')
plt.scatter(xs, ys, c=iris.target)
plt.show()
```

```python
mean = pca.mean_
print(mean)
```

```python
first_pc = pca.components_[0,:]

plt.title('First Principal Component')
plt.scatter(iris.data[:,0], iris.data[:,1])
plt.arrow(mean[0], mean[1], first_pc[0], first_pc[1], color='red', width=0.01)

# Keep axes on same scale
plt.axis('equal')
plt.show()
```

```python
# variance explained by each principal component

from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline

pca = PCA(n_components=2)
pipeline = make_pipeline(StandardScaler(), pca)
pipeline.fit(iris.data)
features = range(pca.n_components_)

plt.bar(features, pca.explained_variance_)
plt.title('Variance Explained by Each PCA Feature (2 components)')
plt.xlabel('PCA feature')
plt.ylabel('variance')
plt.xticks(features)
plt.show()
```

```python
pca = PCA(n_components=3)
pipeline = make_pipeline(StandardScaler(), pca)
pipeline.fit(iris.data)
features = range(pca.n_components_)

plt.bar(features, pca.explained_variance_)
plt.title('Variance Explained by Each PCA Feature (3 components)')
plt.xlabel('PCA feature')
plt.ylabel('variance')
plt.xticks(features)
plt.show()

# TODO - Add vertical line to show where 2 components is 95% variance
# TODO - Add labels for information and noise left and right of the line
```
