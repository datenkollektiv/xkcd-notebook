# NMF-based Recommender System

Non-negative Matrix Factorization (NMF) is a dimension reduction technique that is particularly useful when dealing with sparse data.

```python
import kaggle

# Download a Kaggle dataset
kaggle.api.dataset_download_files('datatattle/covid-19-nlp-text-classification', path='./', unzip=True)
```

```python
import pandas as pd

df = pd.read_csv('Corona_NLP_train.csv', encoding='ISO-8859-1')
```

## NMF rocks interpretable topics

```python
from sklearn.feature_extraction.text import CountVectorizer

countVectorizer = CountVectorizer(stop_words='english', max_df=0.95, min_df=0.05, max_features=1000)
countVector = countVectorizer.fit_transform(df['OriginalTweet'])

print(countVector.shape)
```

```python
from sklearn.decomposition import NMF

nmf = NMF(n_components=10)
nmf_features = nmf.fit_transform(countVector)
```

## How to compare features via "Cosine Similarity"

```python
from sklearn.preprocessing import normalize
norm_features = normalize(nmf_features)

current_tweet = norm_features[0,:]
similarities = norm_features.dot(current_tweet)

print(similarities)
```

```python
# Find similar tweets

import pandas as pd

recommendations = pd.DataFrame(norm_features, index=df['UserName'])

print(recommendations.head())
```

```python
# Grab a random tweet (e.g. first)

current_tweet = recommendations.iloc[0,:]
similarities = recommendations.dot(current_tweet)

# highest similarities
print(similarities.nlargest())
```
