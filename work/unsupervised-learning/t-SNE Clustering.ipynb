# t-SNE t-distributed stochastic neighbor embedding

Maps approximately the local structure ("nearness") of the data in high-dimensional space to a lower-dimensional space. 

```python
# Cluster the iris dataset and run t-SNE on the clusters

from sklearn.datasets import load_iris
iris = load_iris()

import pandas as pd

# Create a dataframe with the four feature variables
df = pd.DataFrame(iris.data, columns=iris.feature_names)
```

```python
from sklearn.manifold import TSNE

model = TSNE(learning_rate=100) # bad choices may lead to unusable results (between 50 and 200 is often good)

transformed = model.fit_transform(df) # the only method of the TSNE model
xs = transformed[:,0]
ys = transformed[:,1]
```

```python
# Visualize the t-SNE (default) results

from matplotlib import pyplot as plt
plt.xkcd()

plt.title('t-SNE Clustering')
plt.scatter(xs, ys, c=iris.target)
plt.show()
```
